```python
ages = [19, 22, 21, 35, 28, 24, 31, 19, 27, 40]

number_of_people = len(ages)
print("Number of people:", number_of_people)

total = sum(ages)
print("Total of all ages:", total)

mean = total / number_of_people
print("Mean age:", round(mean, 1))

youngest = min(ages)
oldest = max(ages)
print("Youngest age:", youngest)
print("Oldest age:", oldest)

ages.sort()

middle1 = ages[4]
middle2 = ages[5]

median = (middle1 + middle2) / 2

print("Median age:", median)
```

```python
import pandas as pd

df3 = pd.read_excel("50_startups_messy.xlsx")

print("Shape (rows, cols):", df3.shape)        
print("\nMissing values per column:")
print(df3.isna().sum())                         
print("\nDuplicate rows:", df3.duplicated().sum())  
print("\nColumn dtypes:")
print(df3.dtypes)                                 

print(df3["State"].unique())
df3 = df3.drop_duplicates()

df3["State"] = df3["State"].str.strip().str.title()

df3["Marketing Spend"] = pd.to_numeric(df3["Marketing Spend"], errors="coerce")

print(df3["State"].unique())
print(df3.dtypes)
print("Remaining missing values:\n", df3.isna().sum())

srs = df3.sample(n=15, random_state=42)
sys_sample = df3.iloc[2::3]
strat = (
    df3.groupby("State", group_keys=False)
       .apply(lambda g: g.sample(frac=0.2, random_state=42))
)
print("Simple random sample size:", len(srs))
print("Systematic sample size:", len(sys_sample))
print("Stratified sample size:", len(strat))


compare = pd.DataFrame({
    "population": df3["State"].value_counts(normalize=True),
    "sample": srs["State"].value_counts(normalize=True),
})
print(compare.round(3))

for col in ["R&D Spend", "Marketing Spend", "Profit"]:
    pop_mean = df3[col].mean()
    sample_mean = srs[col].mean()
    print(f"{col:20s} population mean={pop_mean:>12,.2f}   sample mean={sample_mean:>12,.2f}")
```
